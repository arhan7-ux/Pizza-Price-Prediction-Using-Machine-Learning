```python
import os
import pickle
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# 1. Create a dummy dataset matching your UI fields
data = {
    'Company': ['A', 'A', 'B', 'B'],
    'Diameter (Inches)': [22.0, 12.0, 14.0, 18.0],
    'Topping': ['Chicken', 'Beef', 'Chicken', 'Cheese'],
    'Variant': ['Classic', 'Classic', 'Classic', 'Premium'],
    'Price': [25.0, 12.0, 15.0, 22.0]  # Target variable
}
df = pd.DataFrame(data)

X = df[['Company', 'Diameter (Inches)', 'Topping', 'Variant']]
y = df['Price']

# 2. Set up preprocessing for text columns and build the pipeline
categorical_features = ['Company', 'Topping', 'Variant']
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ],
    remainder='passthrough'
)

model_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', LinearRegression())
])

# 3. Train the model
model_pipeline.fit(X, y)

# 4. Save the model in the current application directory
model_filename = 'pizza_price_model.pkl'
with open(model_filename, 'wb') as file:
    pickle.dump(model_pipeline, file)

print(f"Success! Saved model as: {os.path.abspath(model_filename)}")
```

```python
import os
import pickle
import tkinter as tk
from tkinter import ttk, messagebox
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

MODEL_PATH = "pizza_price_model.pkl"

# ==========================================
# 1. AUTOMATIC MODEL TRAINING & SAVING
# ==========================================
def check_and_train_model():
    """Trains a dummy model if the file doesn't exist."""
    if not os.path.exists(MODEL_PATH):
        print(f"'{MODEL_PATH}' not found. Training a default model...")
        
        # Mock dataset representing companies, sizes, toppings, and price targets
        data = {
            'Company': ['A', 'A', 'B', 'B', 'C', 'C'],
            'Diameter (Inches)': [22.0, 12.0, 14.0, 18.0, 10.0, 16.0],
            'Topping': ['Chicken', 'Beef', 'Chicken', 'Cheese', 'Mushrooms', 'Chicken'],
            'Variant': ['Classic', 'Classic', 'Classic', 'Premium', 'Classic', 'Premium'],
            'Price': [25.50, 12.00, 15.00, 22.00, 9.50, 19.00]
        }
        df = pd.DataFrame(data)
        
        X = df[['Company', 'Diameter (Inches)', 'Topping', 'Variant']]
        y = df['Price']
        
        # Match incoming string features with OneHotEncoding
        preprocessor = ColumnTransformer(
            transformers=[
                ('cat', OneHotEncoder(handle_unknown='ignore'), ['Company', 'Topping', 'Variant'])
            ],
            remainder='passthrough'
        )
        
        pipeline = Pipeline([
            ('prep', preprocessor),
            ('reg', LinearRegression())
        ])
        
        pipeline.fit(X, y)
        
        with open(MODEL_PATH, 'wb') as f:
            pickle.dump(pipeline, f)
        print("Model created and saved successfully.")

# Initialize model generation step
check_and_train_model()

# ==========================================
# 2. TKINTER APPLICATION INTERFACE
# ==========================================
class PizzaPricePredictorApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Pizza Price Predictor")
        self.root.geometry("450x400")
        self.root.configure(pady=20, padx=30)
        
        # Title Label
        title_label = tk.Label(root, text="🍕 Pizza Price Predictor", font=("Arial", 16, "bold"), fg="#b32424")
        title_label.grid(row=0, column=0, columnspan=2, pady=(0, 20))
        
        # Form UI Grid Structure
        # 1. Company
        tk.Label(root, text="Company:", font=("Arial", 11)).grid(row=1, column=0, sticky="w", pady=8)
        self.company_var = tk.StringVar(value="A")
        self.company_menu = ttk.Combobox(root, textvariable=self.company_var, values=["A", "B", "C"], width=22, state="readonly")
        self.company_menu.grid(row=1, column=1, pady=8)
        
        # 2. Diameter
        tk.Label(root, text="Diameter (Inches):", font=("Arial", 11)).grid(row=2, column=0, sticky="w", pady=8)
        self.diameter_entry = ttk.Entry(root, width=24)
        self.diameter_entry.insert(0, "22.0")
        self.diameter_entry.grid(row=2, column=1, pady=8)
        
        # 3. Topping
        tk.Label(root, text="Topping:", font=("Arial", 11)).grid(row=3, column=0, sticky="w", pady=8)
        self.topping_var = tk.StringVar(value="Chicken")
        self.topping_menu = ttk.Combobox(root, textvariable=self.topping_var, values=["Chicken", "Beef", "Cheese", "Mushrooms"], width=22, state="readonly")
        self.topping_menu.grid(row=3, column=1, pady=8)
        
        # 4. Variant
        tk.Label(root, text="Variant:", font=("Arial", 11)).grid(row=4, column=0, sticky="w", pady=8)
        self.variant_var = tk.StringVar(value="Classic")
        self.variant_menu = ttk.Combobox(root, textvariable=self.variant_var, values=["Classic", "Premium"], width=22, state="readonly")
        self.variant_menu.grid(row=4, column=1, pady=8)
        
        # Action Predict Button
        self.predict_btn = tk.Button(root, text="Predict Price", bg="#b32424", fg="white", font=("Arial", 11, "bold"), height=2, width=15, command=self.predict_price)
        self.predict_btn.grid(row=5, column=0, columnspan=2, pady=25)
        
        # Output Result Holder label
        self.result_label = tk.Label(root, text="", font=("Arial", 13, "bold"), fg="green")
        self.result_label.grid(row=6, column=0, columnspan=2)

    def predict_price(self):
        # 1. Safe loading check to verify file exists
        if not os.path.exists(MODEL_PATH):
            messagebox.showerror("Error", f"Model file '{MODEL_PATH}' not found. Please train and save the model first.")
            return

        try:
            # 2. Input Form validation
            diameter_val = float(self.diameter_entry.get().strip())
            
            # 3. Shape input structure back into a DataFrame
            input_df = pd.DataFrame([{
                'Company': self.company_var.get(),
                'Diameter (Inches)': diameter_val,
                'Topping': self.topping_var.get(),
                'Variant': self.variant_var.get()
            }])
            
            # 4. Load binary configuration pipeline
            with open(MODEL_PATH, 'rb') as f:
                model = pickle.load(f)
                
            # 5. Execute internal pipeline steps
            predicted_array = model.predict(input_df)
            final_price = max(0.0, float(predicted_array[0])) # ensure result is not negative
            
            # Display target computation back to screen
            self.result_label.config(text=f"Estimated Price: ${final_price:.2f}", fg="green")
            
        except ValueError:
            messagebox.showerror("Invalid Input", "Please enter a valid numeric value for the pizza diameter.")
        except Exception as e:
            messagebox.showerror("Error", f"An execution error occurred: {str(e)}")

# Driver code execution block
if __name__ == "__main__":
    app_window = tk.Tk()
    app = PizzaPricePredictorApp(app_window)
    app_window.mainloop()
```

```python
import tkinter as tk
from tkinter import messagebox
import joblib
import numpy as np

# Load the pre-trained Machine Learning Model
try:
    model = joblib.load('pizza_price_model.pkl')
except FileNotFoundError:
    model = None

# Mapping dictionaries (Ensure these match your LabelEncoder/OneHotEncoder mappings during training)
company_map = {'A': 0, 'B': 1, 'C': 2, 'D': 3, 'E': 4}
topping_map = {'Chicken': 0, 'Mushrooms': 1, 'Mozzarella': 2, 'Beef': 3, 'Prawns': 4}
variant_map = {'Classic': 0, 'Crunchy': 1, 'Double Deck': 2, 'New York': 3}
size_map = {'Small': 0, 'Medium': 1, 'Large': 2, 'XL': 3}
binary_map = {'No': 0, 'Yes': 1}

# Prediction function
def predict_price():
    if model is None:
        messagebox.showerror("Error", "Model file 'pizza_price_model.pkl' not found. Please train and save the model first.")
        return
        
    try:
        # Extract inputs from GUI components
        company = company_map[company_var.get()]
        diameter = float(diameter_entry.get())
        topping = topping_map[topping_var.get()]
        variant = variant_map[variant_var.get()]
        size = size_map[size_var.get()]
        extra_sauce = binary_map[sauce_var.get()]
        extra_cheese = binary_map[cheese_var.get()]
        
        # Structure features exactly how the ML model expects them
        features = np.array([[company, diameter, topping, variant, size, extra_sauce, extra_cheese]])
        
        # Perform prediction
        predicted_price = model.predict(features)[0]
        
        # Display the result
        result_label.config(text=f"Estimated Price: ${predicted_price:.2f}", fg="#2E7D32")
        
    except ValueError:
        messagebox.showerror("Invalid Input", "Please enter a valid numeric value for Diameter.")
    except Exception as e:
        messagebox.showerror("Error", f"An error occurred: {str(e)}")

# --- UI Setup ---
root = tk.Tk()
root.title("Pizza Price Predictor")
root.geometry("450x550")
root.configure(bg="#F5F5F5")

# Header
header = tk.Label(root, text="🍕 Pizza Price Predictor", font=("Helvetica", 18, "bold"), bg="#F5F5F5", fg="#D32F2F")
header.pack(pady=20)

# Form Frame
frame = tk.Frame(root, bg="#F5F5F5")
frame.pack(pady=10)

def create_dropdown(label_text, options, row):
    lbl = tk.Label(frame, text=label_text, font=("Helvetica", 11), bg="#F5F5F5")
    lbl.grid(row=row, column=0, sticky="w", pady=8, padx=10)
    var = tk.StringVar(root)
    var.set(options[0])
    dropdown = tk.OptionMenu(frame, var, *options)
    dropdown.config(width=18, font=("Helvetica", 10))
    dropdown.grid(row=row, column=1, pady=8, padx=10)
    return var

# Row 0: Company
company_var = create_dropdown("Company:", list(company_map.keys()), 0)

# Row 1: Diameter (Numeric Entry Input)
tk.Label(frame, text="Diameter (Inches):", font=("Helvetica", 11), bg="#F5F5F5").grid(row=1, column=0, sticky="w", pady=8, padx=10)
diameter_entry = tk.Entry(frame, width=21, font=("Helvetica", 10))
diameter_entry.insert(0, "12") # Default value
diameter_entry.grid(row=1, column=1, pady=8, padx=10)

# Row 2: Topping
topping_var = create_dropdown("Topping:", list(topping_map.keys()), 2)

# Row 3: Variant
variant_var = create_dropdown("Variant:", list(variant_map.keys()), 3)

# Row 4: Size
size_var = create_dropdown("Size:", list(size_map.keys()), 4)

# Row 5: Extra Sauce
sauce_var = create_dropdown("Extra Sauce:", list(binary_map.keys()), 5)

# Row 6: Extra Cheese
cheese_var = create_dropdown("Extra Cheese:", list(binary_map.keys()), 6)

# Predict Button
predict_btn = tk.Button(root, text="Predict Price", font=("Helvetica", 12, "bold"), bg="#D32F2F", fg="white", 
                        padx=20, pady=8, command=predict_price, relief="flat")
predict_btn.pack(pady=20)

# Output Label
result_label = tk.Label(root, text="", font=("Helvetica", 16, "bold"), bg="#F5F5F5")
result_label.pack(pady=10)

# Run application
root.mainloop()
```
